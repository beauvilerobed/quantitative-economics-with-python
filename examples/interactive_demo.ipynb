# Interactive Demo: Quantitative Economics Platform

This notebook demonstrates the `qecon_platform` package: an Ornstein–Uhlenbeck
process simulator (useful for simple macroeconomic shocks) and a tiny agent-based
microeconomic simulation. Use the sliders to explore parameter sensitivity.

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider, IntSlider
from qecon_platform.simulation import simulate_ornstein_uhlenbeck, simulate_agent_micro
plt.rcParams.update({'figure.max_open_warning': 0})
```

```python
def plot_ou(theta, mu, sigma, T, dt, n_paths):
    t, paths = simulate_ornstein_uhlenbeck(theta=theta, mu=mu, sigma=sigma, dt=dt, T=T, n_paths=n_paths)
    plt.figure(figsize=(8, 4))
    for i in range(paths.shape[1]):
        plt.plot(t, paths[:, i], lw=1)
    plt.title('Ornstein–Uhlenbeck samples')
    plt.xlabel('time')
    plt.ylabel('value')
    plt.grid(True)
    plt.show()

def plot_agents(n_agents, n_steps, prob_trade):
    wealth = simulate_agent_micro(n_agents=n_agents, n_steps=n_steps, prob_trade=prob_trade)
    mean_wealth = wealth.mean(axis=1)
    plt.figure(figsize=(8, 4))
    plt.plot(mean_wealth)
    plt.title('Average wealth over time')
    plt.xlabel('time step')
    plt.ylabel('average wealth')
    plt.grid(True)
    plt.show()

# Interactive controls
interact(plot_ou,
         theta=FloatSlider(value=0.5, min=0.0, max=2.0, step=0.1),
         mu=FloatSlider(value=0.0, min=-1.0, max=1.0, step=0.1),
         sigma=FloatSlider(value=0.1, min=0.0, max=1.0, step=0.01),
         T=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1),
         dt=FloatSlider(value=0.01, min=0.001, max=0.1, step=0.001),
         n_paths=IntSlider(value=3, min=1, max=20));

interact(plot_agents,
         n_agents=IntSlider(value=100, min=10, max=1000, step=10),
         n_steps=IntSlider(value=200, min=10, max=2000, step=10),
         prob_trade=FloatSlider(value=0.1, min=0.0, max=1.0, step=0.01));
```
